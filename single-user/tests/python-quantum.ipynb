# qiskit - sampler

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

N = 1024
sampler = StatevectorSampler()
result_stats = sampler.run([qc], shots=N).result()
result_counts = result_stats[0].data.meas.get_counts()
print(result_counts)
result_count = sum(result_counts.values())
if result_count != N:
    raise(AssertionError(f"Sampler result count is {result_count}, expected {N}"))
plot_histogram(result_counts)
```

# PennyLane

```python
import numpy
import pennylane as qp

def qfunc(x, y):
    qp.RZ(x, wires=0)
    qp.CNOT(wires=[0,1])
    qp.RY(y, wires=1)
    return qp.expval(qp.Z(wires=1))

dev = qp.device('default.qubit', wires=2)
circuit = qp.QNode(qfunc, dev)
fig, ax = qp.draw_mpl(circuit)(numpy.pi/4, 0.7)
```
